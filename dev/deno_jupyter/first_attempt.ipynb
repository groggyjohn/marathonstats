## Finding an entry

```typescript
function findEntry(y, g, c) {
    return histogramData.filter(row => row.year === y && row.gender === g && row.category === c )[0];
}
```

```typescript
const year = 2026;
const gender = 'male';
const category = '18-39';

console.log(findEntry(year, gender, category).count);
```
